# Ч. 2, №10

**Постановка:**
$$ J(x) = \int_{0}^{\pi} (\dot{x}^2 - x^2) dt \rightarrow \text{extr} $$
$$ x(0) = 0, \quad \int_{0}^{\pi} x \cos t \, dt = \frac{\pi}{2}, \quad \int_{0}^{\pi} x \sin t \, dt = -2 $$

**Замена $\dot{x} = u$, введение $y_1, y_2$:**
$$ y_1(t) = \int_{0}^{t} x \cos \tau \, d\tau, \quad y_2(t) = \int_{0}^{t} x \sin \tau \, d\tau $$
$$ \mathcal{L} = \int_{0}^{\pi} [\lambda_{0}(u^{2} - x^{2}) + p(\dot{x} - u) + q_{1}(\dot{y}_{1} - x \cos t) + q_{2}(\dot{y}_{2} - x \sin t)] dt + l $$
$$ l = \lambda_{1}x(0) + \lambda_{2}y_{1}(0) + \lambda_{3}y_{2}(0) + \lambda_{4}(y_{1}(\pi) - \frac{\pi}{2}) + \lambda_{5}(y_{2}(\pi) + 2) $$

**Функция Гамильтона:**
$$ H = pu + q_1x \cos t + q_2x \sin t - \lambda_0(u^2 - x^2) $$

## 1-4. Условия ПМП и нормальность

**1. Стационарность по \(u\) (условие максимума ПМП):**
$$ \frac{\partial H}{\partial u} = p - 2\lambda_0 u = 0 \quad \Rightarrow \quad p = 2\lambda_0 u $$
$$ \frac{\partial^2 H}{\partial u^2} = -2\lambda_0 \le 0 \quad (\text{для минимума}) $$

**2. Стационарность по переменным \((x, y_1, y_2)\):**

*   **По \(x\):**
    $$ \frac{\partial H}{\partial x} = q_1 \cos t + q_2 \sin t + 2\lambda_0 x, \quad \dot{p} = - \frac{\partial H}{\partial x} \quad \Rightarrow \quad \dot{p} = -2\lambda_0 x - q_1 \cos t - q_2 \sin t $$
*   **По \(y_1\):**
    $$ \frac{\partial H}{\partial y_1} = 0, \quad \dot{q}_1 = -\frac{\partial H}{\partial y_1} \quad \Rightarrow \quad \dot{q}_1 = 0 \quad \Rightarrow \quad q_1 = \text{const} $$
*   **По \(y_2\):**
    $$ \frac{\partial H}{\partial y_2} = 0, \quad \dot{q}_2 = -\frac{\partial H}{\partial y_2} \quad \Rightarrow \quad \dot{q}_2 = 0 \quad \Rightarrow \quad q_2 = \text{const} $$

**3. Условия трансверсальности:**
$$ p(0) = \frac{\partial l}{\partial x(0)} = \lambda_1, \quad p(\pi) = -\frac{\partial l}{\partial x(\pi)} = 0 $$
$$ q_1(0) = \frac{\partial l}{\partial y_1(0)} = \lambda_2, \quad q_1(\pi) = -\frac{\partial l}{\partial y_1(\pi)} = -\lambda_4 $$
$$ q_2(0) = \frac{\partial l}{\partial y_2(0)} = \lambda_3, \quad q_2(\pi) = -\frac{\partial l}{\partial y_2(\pi)} = -\lambda_5 $$

**4. Нормальность и знаки:**

$$ \lambda_0 \ge 0 \text{ (loc min)} $$

При $\lambda_0 = 0 \Rightarrow p = 0 \Rightarrow q_1 = q_2 = 0 \Rightarrow \lambda_1 = \dots = \lambda_5 = 0$ (тривиально). Следовательно, $\lambda_0 = 1$.

$$ p = 2u = 2\dot{x} $$

## 8-9. Решение системы и экстремаль

**8. Решение системы:**
$$ \begin{cases} -\frac{\pi}{4}a + \frac{\pi^2}{4}b = \frac{\pi}{2} \\ C_2 \frac{\pi}{2} + \frac{\pi^2}{4}a - \frac{\pi}{4}b = -2 \\ C_2 + \pi a + b = 0 \end{cases} \Rightarrow \begin{cases} -a + \pi b = 2 \\ 2\pi C_2 + \pi^2 a - \pi b = -8 \\ C_2 = -\pi a - b \end{cases} $$
$$ -\pi^2 a - 3\pi b = -8 \quad \Rightarrow \quad -\pi^3 b + 2\pi^2 - 3\pi b = -8 $$
$$ \pi(\pi^2 + 3)b = 2(\pi^2 + 4) \Rightarrow b = \frac{2(\pi^2 + 4)}{\pi(\pi^2 + 3)} $$
$$ a = \pi b - 2 = \frac{2}{\pi^2 + 3}, \quad C_2 = -\pi a - b = -\frac{4(\pi^2 + 2)}{\pi(\pi^2 + 3)} $$

**9. Экстремаль:**
$$ \hat{x}(t) = \frac{2}{\pi(3 + \pi^2)} [-2(2 + \pi^2) \sin t + t((4 + \pi^2) \cos t + \pi \sin t)] $$

## 10-12. Достаточность, максимум и ответ

**10. Глобальный минимум:**
$$ x = \hat{x} + h, \quad h(0) = 0, \quad \int_{0}^{\pi} h \cos t \, dt = \int_{0}^{\pi} h \sin t \, dt = 0 $$
$$ J(\hat{x} + h) - J(\hat{x}) = \int_{0}^{\pi} (2\hat{x}\dot{h} - 2\hat{x}h)dt + \int_{0}^{\pi} (\dot{h}^2 - h^2)dt $$
$$ 2 \int_{0}^{\pi} \hat{x} \dot{h} \, dt = -2 \int_{0}^{\pi} \ddot{x} h \, dt \Rightarrow -2 \int_{0}^{\pi} (\ddot{x} + \hat{x}) h \, dt = \int_{0}^{\pi} (q_1 \cos t + q_2 \sin t) h \, dt = 0 $$
$$ \Delta J = \int_{0}^{\pi} (\dot{h}^2 - h^2) dt, \quad v = \dot{h} \perp \{\sin t, 1 + \cos t\} $$
$$ \int_{0}^{\pi} h^2 dt \le C \int_{0}^{\pi} v^2 dt, \quad C = \frac{5\pi^4 - 84\pi^2 + 384}{6(3\pi^2 - 16)} \approx 0.5144 < 1 $$
$$ \Delta J \ge (1 - C) \int_{0}^{\pi} v^2 dt > 0 \quad (v \not\equiv 0) \Rightarrow \hat{x} \text{ — строгий глобальный минимум} $$

**11. Отсутствие максимума:**
$$ H^{\max} = u^2 - x^2 + pu + \dots \Rightarrow H_{uu}^{\max} = 2 > 0 \Rightarrow \sup_u H^{\max} = +\infty $$
$$ h = \alpha \sin 3t \Rightarrow h(0) = 0, \quad \int_{0}^{\pi} h \cos t \, dt = 0, \quad \int_{0}^{\pi} h \sin t \, dt = 0 $$
$$ \Delta J = \int_{0}^{\pi} (9\alpha^2 \cos^2 3t - \alpha^2 \sin^2 3t) dt = \alpha^2 \left(9 \cdot \frac{\pi}{2} - \frac{\pi}{2}\right) = 4\pi\alpha^2 $$
$$ \lim_{|\alpha| \to \infty} J(\hat{x} + h) = J_{\min} + 4\pi\alpha^2 = +\infty \Rightarrow \text{ глобального максимума нет.} $$

**12. Значение минимума:**

$$ J(\hat{x}) = \int_{0}^{\pi} (\dot{\hat{x}}^2 - \hat{x}^2) dt = [\hat{x}\dot{\hat{x}}]_0^{\pi} - \int_{0}^{\pi} \hat{x}(\ddot{\hat{x}} + \hat{x}) dt = 0 - \int_{0}^{\pi} \hat{x} \left(-\frac{q_1}{2} \cos t - \frac{q_2}{2} \sin t\right) dt $$

$$ J(\hat{x}) \equiv \frac{q_1}{2} \int_0^\pi \hat{x} \cos t \, dt + \frac{q_2}{2} \int_0^\pi \hat{x} \sin t \, dt = \frac{q_1}{2} \cdot \frac{\pi}{2} + \frac{q_2}{2} \cdot (-2) = \frac{\pi q_1}{4} - q_2 $$

Подставляя $q_1 = -4a$ и $q_2 = 4b$:

$$ J_{\min} = -\pi a - 4b = -\pi \left( \frac{2}{\pi^2 + 3} \right) - 4 \left( \frac{2(\pi^2 + 4)}{\pi(\pi^2 + 3)} \right) = -\frac{2(5\pi^2 + 16)}{\pi(\pi^2 + 3)} $$

---
---
### Ответ

$$ \hat{x}(t) = \frac{2}{\pi(3 + \pi^2)} [-2(2 + \pi^2) \sin t + t((4 + \pi^2) \cos t + \pi \sin t)] $$

$$ \hat{u}(t) = \frac{2}{\pi(3 + \pi^2)} [-\pi^2 \cos t + \pi \sin t + t(\pi \cos t - (4 + \pi^2) \sin t)] $$

$$ J_{\min} = -\frac{2(5\pi^2 + 16)}{\pi(3 + \pi^2)} \approx -3.2326 $$

**Характер экстремума:**
*   Траектория $\hat{x}(t)$ доставляет **строгий глобальный минимум** (доказано через положительность второй вариации: $\Delta J \ge (1-C) \int \dot{h}^2 dt > 0$).
*   **Глобального максимума не существует**, так как для задачи на максимум функция Гамильтона выпукла по $u$ ($H_{uu}^{\max} = 2 > 0$), и при вариации $h(t) = \alpha \sin 3t$ имеем $J \to +\infty$ при $|\alpha| \to \infty$.

```python
import matplotlib
matplotlib.use('Agg') 
import warnings
warnings.filterwarnings("ignore") # Убираем предупреждения библиотек

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import Image, display
import scipy.integrate as integrate

# --- 1. Определение математических функций ---
def x_opt(t):
    pi = np.pi
    return (2 / (pi * (3 + pi**2))) * (-2 * (2 + pi**2) * np.sin(t) + t * ((4 + pi**2) * np.cos(t) + pi * np.sin(t)))

def u_opt(t):
    pi = np.pi
    return (2 / (pi * (3 + pi**2))) * (-pi**2 * np.cos(t) + pi * np.sin(t) + t * (pi * np.cos(t) - (4 + pi**2) * np.sin(t)))

def u_prime(t):
    pi = np.pi
    q1 = -8 / (pi**2 + 3)
    q2 = 8 * (pi**2 + 4) / (pi * (pi**2 + 3))
    return -x_opt(t) - (q1/2)*np.cos(t) - (q2/2)*np.sin(t)

def L_func(t):
    return u_opt(t)**2 - x_opt(t)**2

# --- 2. Подготовка данных (оптимизировано для скорости) ---
# 35 кадров достаточно для плавности и работает в 2 раза быстрее
t_vals = np.linspace(0, np.pi, 35) 
x_vals = x_opt(t_vals)
u_vals = u_opt(t_vals)
L_vals = L_func(t_vals)
J_vals = [integrate.quad(L_func, 0, t)[0] for t in t_vals]

# Промежуточные точки (каждые pi/8)
t_checkpoints = np.linspace(0, np.pi, 9)
x_checkpoints = x_opt(t_checkpoints)
u_checkpoints = u_opt(t_checkpoints)

# --- 3. Настройка фигуры и графиков ---
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('Анимация оптимального процесса', fontsize=16, fontweight='bold')
plt.subplots_adjust(hspace=0.4, wspace=0.3)

# 1. Траектория
ax1.set_xlim(0, np.pi); ax1.set_ylim(min(x_vals)-0.3, 0.2); ax1.grid(True, alpha=0.5); ax1.set_title(r'1. Траектория \(x(t)\)')
ax1.plot(t_checkpoints, x_checkpoints, 'ko', ms=4, alpha=0.5) 
line1, = ax1.plot([], [], 'b-', lw=2)
dot1, = ax1.plot([], [], 'ro', ms=8)
txt1 = ax1.text(0.05, 0.1, '', transform=ax1.transAxes, fontsize=10, bbox=dict(facecolor='white', alpha=0.8, edgecolor='gray'))

# 2. Управление
ax2.set_xlim(0, np.pi); ax2.set_ylim(min(u_vals)-0.2, max(u_vals)+0.2); ax2.grid(True, alpha=0.5); ax2.set_title(r'2. Управление \(u(t)\)')
ax2.plot(t_checkpoints, u_checkpoints, 'ko', ms=4, alpha=0.5)
line2, = ax2.plot([], [], 'r-', lw=2)
dot2, = ax2.plot([], [], 'bo', ms=8)
txt2 = ax2.text(0.05, 0.8, '', transform=ax2.transAxes, fontsize=10, bbox=dict(facecolor='white', alpha=0.8, edgecolor='gray'))

# 3. Фазовый портрет со стрелкой
ax3.set_xlim(min(x_vals)-0.3, max(x_vals)+0.3); ax3.set_ylim(min(u_vals)-0.2, max(u_vals)+0.2); ax3.grid(True, alpha=0.5); ax3.set_title(r'3. Фазовый портрет \((x, u)\)')
ax3.plot(x_checkpoints, u_checkpoints, 'ko', ms=4, alpha=0.5) 
line3, = ax3.plot([], [], 'g-', lw=2)
dot3, = ax3.plot([], [], 'ro', ms=8)
quiver = ax3.quiver([], [], [], [], color='black', angles='xy', scale_units='xy', scale=1, width=0.008)

# 4. Накопление площади
ax4.set_xlim(0, np.pi); ax4.set_ylim(min(L_vals)-1, 1); ax4.grid(True, alpha=0.5); ax4.set_title(r'4. Накопление площади \(J(t)\)')
line4, = ax4.plot([], [], 'm-', lw=2)
dot4, = ax4.plot([], [], 'ko', ms=8)
txt4 = ax4.text(0.05, 0.1, '', transform=ax4.transAxes, fontsize=10, bbox=dict(facecolor='white', alpha=0.8, edgecolor='gray'))
fill4 = None

# --- 4. Функция анимации ---
def animate(i):
    global fill4
    # Обновляем линии и точки
    line1.set_data(t_vals[:i], x_vals[:i]); dot1.set_data([t_vals[i]], [x_vals[i]]); txt1.set_text(f't={t_vals[i]:.2f}, x={x_vals[i]:.2f}')
    line2.set_data(t_vals[:i], u_vals[:i]); dot2.set_data([t_vals[i]], [u_vals[i]]); txt2.set_text(f't={t_vals[i]:.2f}, u={u_vals[i]:.2f}')
    line3.set_data(x_vals[:i], u_vals[:i]); dot3.set_data([x_vals[i]], [u_vals[i]])
    
    # Обновляем стрелку на фазовом портрете
    quiver.set_offsets([[x_vals[i], u_vals[i]]])
    quiver.set_UVC(u_vals[i]*0.4, u_prime(t_vals[i])*0.4) 
    
    # Обновляем площадь
    line4.set_data(t_vals[:i], L_vals[:i]); dot4.set_data([t_vals[i]], [L_vals[i]]); txt4.set_text(f'J={J_vals[i]:.2f}')
    if fill4 is not None:
        fill4.remove()
    fill4 = ax4.fill_between(t_vals[:i], 0, L_vals[:i], color='yellow', alpha=0.5)
    
    return line1, dot1, line2, dot2, line3, dot3, quiver, line4, dot4, txt1, txt2, txt4

# --- 5. Генерация и отображение GIF ---
print("Генерация GIF... Пожалуйста, подождите около 10-15 секунд.")
# interval=100 и fps=10 работают синхронно для плавного и быстрого результата
anim = FuncAnimation(fig, animate, frames=len(t_vals), interval=100, blit=False, repeat=False)

filename = 'optimization_anim.gif'
anim.save(filename, writer=PillowWriter(fps=10))
plt.close(fig) 

print("Готово! Анимация ниже:")
display(Image(filename=filename))
```
