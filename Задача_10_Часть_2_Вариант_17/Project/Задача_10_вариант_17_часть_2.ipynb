# Задача 10 (Часть 2) — Вариант 17

## 1. Постановка задачи

$$
\begin{aligned}
& J[x]=\int_0^\pi(\dot x^2-x^2)\,dt\to\operatorname{extr}, \\
& x(0)=0,\qquad \int_0^\pi x\cos t\,dt=\frac\pi2,\qquad \int_0^\pi x\sin t\,dt=-2.
\end{aligned}
$$

## 2. Сведение к задаче Лагранжа

Введем замену переменной \($\dot{x} = u$;\). Тогда функционал примет вид:

$$
\int_0^\pi (u^2 - x^2)\,dt \to \operatorname{extr},
$$

с дополнительной связью и граничными условиями:

$$
\dot{x} - u = 0, \quad x(0) = 0, \quad \int_0^\pi x \cos t \, dt = \frac{\pi}{2}, \quad \int_0^\pi x \sin t \, dt = -2.
$$

Функция Лагранжа:

$$
\begin{aligned}
& L = \lambda_0(u^2 - x^2) + p(t)(\dot{x} - u) - \lambda_1 x \cos t - \lambda_2 x \sin t, \\
& l = \lambda_1 \frac{\pi}{2} - 2\lambda_2 + \lambda_3 x(0).
\end{aligned}
$$

Множители: \($\lambda_0, \lambda_1, \lambda_2, \lambda_3$\) и $\(p(t)\$).

## 3. Необходимые условия

1. **Стационарность по \(x\):**
$$
\frac{d}{dt} L_{\dot{x}} = L_x \implies \dot{p} = -2\lambda_0 x - \lambda_1 \cos t - \lambda_2 \sin t.
$$

2. **Стационарность по \(u\):**
$$
L_u = 0 \implies p = 2\lambda_0 u.
$$

3. **Условия трансверсальности:**
$$
p(0) = \lambda_3, \quad p(\pi) = 0.
$$

4. **Согласование знаков:** \($\lambda_0 \ge 0$\); \($\lambda_1, \lambda_2, \lambda_3$\) любого знака.

## 4. Исследование нормальности

**Случай \($\lambda_0 = 0$\):**
$$
\begin{aligned}
& p \equiv 0 \implies \lambda_1 \cos t + \lambda_2 \sin t \equiv 0 \implies \lambda_1 = \lambda_2 = 0 \\
& \implies \lambda_3 = 0 \implies \text{все множители нулевые — невозможно.}
\end{aligned}
$$

**Случай \($\lambda_0 \neq 0$\):**
$$
\begin{aligned}
& \lambda_0 = 1 \implies p = 2\dot{x}, \quad \ddot{x} + x = -\frac{\lambda_1}{2} \cos t - \frac{\lambda_2}{2} \sin t, \\
& x(0) = 0, \quad \dot{x}(\pi) = 0.
\end{aligned}
$$

## 5. Решение краевой задачи

$$
\begin{aligned}
& \ddot{x} + x = -\frac{\lambda_1}{2} \cos t - \frac{\lambda_2}{2} \sin t \equiv 2a \cos t - 2b \sin t, \\
& a = -\frac{\lambda_1}{4}, \quad b = \frac{\lambda_2}{4} \\
& x(t) = C_1 \cos t + C_2 \sin t + at \sin t + bt \cos t \\
& x(0) = 0 \implies C_1 = 0 \\
& \dot{x}(t) = C_2 \cos t + a(\sin t + t \cos t) + b(\cos t - t \sin t) \\
& \dot{x}(\pi) = 0 \implies -C_2 - \pi a - b = 0 \implies C_2 = -\pi a - b
\end{aligned}
$$

## 6. Интегральные условия

$$
\int_0^\pi \sin t \cos t \, dt = 0, \qquad \int_0^\pi t \sin t \cos t \, dt = -\frac{\pi}{4},
$$
$$
\int_0^\pi \sin^2 t \, dt = \frac{\pi}{2}, \qquad \int_0^\pi t \cos^2 t \, dt = \int_0^\pi t \sin^2 t \, dt = \frac{\pi^2}{4}.
$$

Подставляем \(x(t)\) в интегральные ограничения:
$$
\begin{aligned}
& \int_0^\pi x \cos t \, dt = -\frac{\pi}{4}a + \frac{\pi^2}{4}b = \frac{\pi}{2} \quad \Rightarrow \quad -a + \pi b = 2. \\
& \int_0^\pi x \sin t \, dt = \frac{\pi}{2}C_2 + \frac{\pi^2}{4}a - \frac{\pi}{4}b = -2.
\end{aligned}
$$

Подставим \($C_2 = -\pi a - b$\):
$$
\pi^2 a + 3\pi b = 8.
$$

Из первого уравнения \($a = \pi b - 2$\). Подставляем:
$$
\pi(\pi^2 + 3)b = 2(\pi^2 + 4) \implies b = \frac{2(\pi^2 + 4)}{\pi(\pi^2 + 3)}, \quad a = \frac{2}{\pi^2 + 3}.
$$

Следовательно:
$$
C_2 = -\frac{4(\pi^2 + 2)}{\pi(\pi^2 + 3)}.
$$

## 7. Множители и экстремаль

$$
\begin{aligned}
& \lambda_0 = 1, \quad \lambda_1 = -4a = -\frac{8}{\pi^2 + 3}, \quad \lambda_2 = 4b = \frac{8(\pi^2 + 4)}{\pi(\pi^2 + 3)}, \\
& \lambda_3 = p(0) = 2\dot{x}(0) = -\frac{4\pi}{\pi^2 + 3}.
\end{aligned}
$$

Единственная допустимая экстремаль:
$$
\hat{x}(t) = \frac{2}{\pi(\pi^2 + 3)} \left[ -2(\pi^2 + 2) \sin t + \pi t \sin t + (\pi^2 + 4) t \cos t \right].
$$

Вычисляем оптимальное управление \($\hat{u}(t)$ = $\dot{\hat{x}}(t)$\), дифференцируя каждое слагаемое в скобках:
$$
\begin{aligned}
& \frac{d}{dt} \bigl[-2(\pi^2+2)\sin t\bigr] = -2(\pi^2+2)\cos t. \\
& \frac{d}{dt} \bigl[\pi\,t\sin t\bigr] = \pi(\sin t + t\cos t). \\
& \frac{d}{dt} \bigl[(\pi^2+4)\,t\cos t\bigr] = (\pi^2+4)(\cos t - t\sin t).
\end{aligned}
$$

Группируем полученные слагаемые:
$$
\hat{u}(t) = \frac{2}{\pi(\pi^2+3)} \Bigl[ \underbrace{-2(\pi^2+2)\cos t + (\pi^2+4)\cos t}_{-\pi^2 \cos t} + \pi\sin t + t\bigl(\pi\cos t - (\pi^2+4)\sin t\bigr) \Bigr].
$$

Итоговое управление:
$$
\hat{u}(t) = \frac{2}{\pi(\pi^2 + 3)} \left[ -\pi^2 \cos t + \pi \sin t + t(\pi \cos t - (\pi^2 + 4) \sin t) \right].
$$

Значение функционала:
$$
J(\hat{x}) = -\pi a - 4b = -\frac{2(5\pi^2 + 16)}{\pi(\pi^2 + 3)} \approx -3{,}2326.
$$

## 8. Исследование достаточности

$$
\Delta J(h) = \int_0^\pi (\dot{h}^2 - h^2) \, dt.
$$

1. **Условие Лежандра:**
$$
F_{\dot{x}\dot{x}} = 2 > 0 \implies \text{кандидат на строгий локальный минимум.}
$$

2. **Условие Якоби.**
$$
\ddot{y} + y = 0, \quad y(0)=0, \quad \dot{y}(0)=1 \implies y(t) = \sin t.
$$
$$
y(\pi) = 0 \implies t^* = \pi \text{ — сопряжённая точка.}
$$
* Обычное условие Якоби на \($0,\pi$\): **выполнено**.
* Усиленное условие Якоби на \($0,\pi]$\: **не выполнено**.

Без ограничений: \($h = \sin t \implies \Delta J(\sin t) = 0$\).

Для допустимых вариаций:
$$
\int_0^\pi h\cos t \, dt = 0, \quad \int_0^\pi h\sin t \, dt = 0 \implies \sin t \notin \{h\} \implies \Delta J(h) > 0 \quad (h \not\equiv 0).
$$

3. **Проверка вариацией.** Возьмем \($h = \sin 3t$\):
$$
\int_0^\pi \sin 3t\cos t \, dt = 0, \quad \int_0^\pi \sin 3t\sin t \, dt = 0 \quad (\text{допустима}).
$$
$$
\Delta J(\sin 3t) = \int_0^\pi (9\cos^2 3t - \sin^2 3t) \, dt = 4\pi > 0.
$$
$$
J(\hat{x} + \mu \sin 3t) = J(\hat{x}) + 4\pi\mu^2 \xrightarrow{\mu\to\infty} +\infty \implies \text{глобального максимума нет.}
$$

**Вывод:** \($\Delta J(h) > 0$\) для любых допустимых \($h \not\equiv 0 \implies \hat{x}(t)$\) доставляет **строгий глобальный минимум**.

## 9. Строгое доказательство через неравенство Виртингера

Для \($h \in C^1[0,\pi],\ h(0)=0$\) справедливо неравенство Виртингера:
$$
\int_0^\pi h^2(t)\,dt \le \int_0^\pi \dot{h}^2(t)\,dt,
$$
причем равенство достигается тогда и только тогда, когда:
$$
h(t) = C\sin t, \quad C = \text{const}.
$$

Запишем вторую вариацию:
$$
\Delta J(h) = \int_0^\pi \dot{h}^2(t)\,dt - \int_0^\pi h^2(t)\,dt \ge 0.
$$

Предположим, что достигается равенство \($\Delta J(h) = 0$\):
$$
\Delta J(h) = 0 \implies h(t) = C\sin t.
$$

Проверим выполнение интегрального ограничения \($\int_0^\pi h(t)\sin t\,dt = 0$\) для такой вариации:
$$
\int_0^\pi (C\sin t)\sin t\,dt = C\int_0^\pi \sin^2 t\,dt = C\frac{\pi}{2}.
$$

По условию \($C\frac{\pi}{2}$ = 0 $\implies C = 0$\). 

Отсюда:
$$
C=0 \implies h(t) \equiv 0.
$$

Таким образом, равенство \($\Delta J(h) = 0$\) возможно только для нулевой вариации:
$$
h \not\equiv 0 \implies \Delta J(h) > 0.
$$

## Ответ

$$
\hat{x}(t) = \frac{2}{\pi(\pi^2 + 3)} \left[ -2(\pi^2 + 2) \sin t + \pi t \sin t + (\pi^2 + 4) t \cos t \right],
$$

$$
\hat{u}(t) = \frac{2}{\pi(\pi^2 + 3)} \left[ -\pi^2 \cos t + \pi \sin t + t(\pi \cos t - (\pi^2 + 4) \sin t) \right],
$$

$$
J_{\min} = -\frac{2(5\pi^2 + 16)}{\pi(\pi^2 + 3)} \approx -3{,}2326.
$$

Траектория \($\hat{x}(t)$\) доставляет **строгий глобальный минимум**; глобального максимума не существует.

```python
import matplotlib
matplotlib.use('Agg') 
import warnings
warnings.filterwarnings("ignore") # Убираем предупреждения библиотек

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import Image, display
import scipy.integrate as integrate

# --- 1. Определение математических функций ---
def x_opt(t):
    pi = np.pi
    return (2 / (pi * (3 + pi**2))) * (-2 * (2 + pi**2) * np.sin(t) + t * ((4 + pi**2) * np.cos(t) + pi * np.sin(t)))

def u_opt(t):
    pi = np.pi
    return (2 / (pi * (3 + pi**2))) * (-pi**2 * np.cos(t) + pi * np.sin(t) + t * (pi * np.cos(t) - (4 + pi**2) * np.sin(t)))

def u_prime(t):
    pi = np.pi
    q1 = -8 / (pi**2 + 3)
    q2 = 8 * (pi**2 + 4) / (pi * (pi**2 + 3))
    return -x_opt(t) - (q1/2)*np.cos(t) - (q2/2)*np.sin(t)

def L_func(t):
    return u_opt(t)**2 - x_opt(t)**2

# --- 2. Подготовка данных (оптимизировано для скорости) ---
# 35 кадров достаточно для плавности и работает в 2 раза быстрее
t_vals = np.linspace(0, np.pi, 35) 
x_vals = x_opt(t_vals)
u_vals = u_opt(t_vals)
L_vals = L_func(t_vals)
J_vals = [integrate.quad(L_func, 0, t)[0] for t in t_vals]

# Промежуточные точки (каждые pi/8)
t_checkpoints = np.linspace(0, np.pi, 9)
x_checkpoints = x_opt(t_checkpoints)
u_checkpoints = u_opt(t_checkpoints)

# --- 3. Настройка фигуры и графиков ---
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('Анимация оптимального процесса', fontsize=16, fontweight='bold')
plt.subplots_adjust(hspace=0.4, wspace=0.3)

# 1. Траектория
ax1.set_xlim(0, np.pi); ax1.set_ylim(min(x_vals)-0.3, 0.2); ax1.grid(True, alpha=0.5); ax1.set_title(r'1. Траектория \(x(t)\)')
ax1.plot(t_checkpoints, x_checkpoints, 'ko', ms=4, alpha=0.5) 
line1, = ax1.plot([], [], 'b-', lw=2)
dot1, = ax1.plot([], [], 'ro', ms=8)
txt1 = ax1.text(0.05, 0.1, '', transform=ax1.transAxes, fontsize=10, bbox=dict(facecolor='white', alpha=0.8, edgecolor='gray'))

# 2. Управление
ax2.set_xlim(0, np.pi); ax2.set_ylim(min(u_vals)-0.2, max(u_vals)+0.2); ax2.grid(True, alpha=0.5); ax2.set_title(r'2. Управление \(u(t)\)')
ax2.plot(t_checkpoints, u_checkpoints, 'ko', ms=4, alpha=0.5)
line2, = ax2.plot([], [], 'r-', lw=2)
dot2, = ax2.plot([], [], 'bo', ms=8)
txt2 = ax2.text(0.05, 0.8, '', transform=ax2.transAxes, fontsize=10, bbox=dict(facecolor='white', alpha=0.8, edgecolor='gray'))

# 3. Фазовый портрет со стрелкой
ax3.set_xlim(min(x_vals)-0.3, max(x_vals)+0.3); ax3.set_ylim(min(u_vals)-0.2, max(u_vals)+0.2); ax3.grid(True, alpha=0.5); ax3.set_title(r'3. Фазовый портрет \((x, u)\)')
ax3.plot(x_checkpoints, u_checkpoints, 'ko', ms=4, alpha=0.5) 
line3, = ax3.plot([], [], 'g-', lw=2)
dot3, = ax3.plot([], [], 'ro', ms=8)
quiver = ax3.quiver([], [], [], [], color='black', angles='xy', scale_units='xy', scale=1, width=0.008)

# 4. Накопление площади
ax4.set_xlim(0, np.pi); ax4.set_ylim(min(L_vals)-1, 1); ax4.grid(True, alpha=0.5); ax4.set_title(r'4. Накопление площади \(J(t)\)')
line4, = ax4.plot([], [], 'm-', lw=2)
dot4, = ax4.plot([], [], 'ko', ms=8)
txt4 = ax4.text(0.05, 0.1, '', transform=ax4.transAxes, fontsize=10, bbox=dict(facecolor='white', alpha=0.8, edgecolor='gray'))
fill4 = None

# --- 4. Функция анимации ---
def animate(i):
    global fill4
    # Обновляем линии и точки
    line1.set_data(t_vals[:i], x_vals[:i]); dot1.set_data([t_vals[i]], [x_vals[i]]); txt1.set_text(f't={t_vals[i]:.2f}, x={x_vals[i]:.2f}')
    line2.set_data(t_vals[:i], u_vals[:i]); dot2.set_data([t_vals[i]], [u_vals[i]]); txt2.set_text(f't={t_vals[i]:.2f}, u={u_vals[i]:.2f}')
    line3.set_data(x_vals[:i], u_vals[:i]); dot3.set_data([x_vals[i]], [u_vals[i]])
    
    # Обновляем стрелку на фазовом портрете
    quiver.set_offsets([[x_vals[i], u_vals[i]]])
    quiver.set_UVC(u_vals[i]*0.4, u_prime(t_vals[i])*0.4) 
    
    # Обновляем площадь
    line4.set_data(t_vals[:i], L_vals[:i]); dot4.set_data([t_vals[i]], [L_vals[i]]); txt4.set_text(f'J={J_vals[i]:.2f}')
    if fill4 is not None:
        fill4.remove()
    fill4 = ax4.fill_between(t_vals[:i], 0, L_vals[:i], color='yellow', alpha=0.5)
    
    return line1, dot1, line2, dot2, line3, dot3, quiver, line4, dot4, txt1, txt2, txt4

# --- 5. Генерация и отображение GIF ---
print("Генерация GIF... Пожалуйста, подождите около 10-15 секунд.")
# interval=100 и fps=10 работают синхронно для плавного и быстрого результата
anim = FuncAnimation(fig, animate, frames=len(t_vals), interval=100, blit=False, repeat=False)

filename = 'optimization_anim.gif'
anim.save(filename, writer=PillowWriter(fps=10))
plt.close(fig) 

print("Готово! Анимация ниже:")
display(Image(filename=filename))
```
