# Вариант 17, Задача 2

## Пункт 1

$$F(x,y) = (xy,\ x^2+y^2,\ |x|),\qquad z=(x,y),\qquad h=(h_1,h_2)$$

**Шаг 1.** Приращение аргумента:
$$z + \lambda h = (x+\lambda h_1,\ y+\lambda h_2)$$

**Шаг 2.** Подстановка в $F$:
$$F(z+\lambda h) = \Big((x+\lambda h_1)(y+\lambda h_2),\ (x+\lambda h_1)^2+(y+\lambda h_2)^2,\ |x+\lambda h_1|\Big)$$

**Шаг 3.** Раскрытие скобок:
$$
\begin{aligned}
1\text{-я:} &\quad xy + \lambda(yh_1 + xh_2) + \lambda^2 h_1 h_2 \\
2\text{-я:} &\quad x^2 + y^2 + \lambda(2xh_1 + 2yh_2) + \lambda^2(h_1^2+h_2^2) \\
3\text{-я:} &\quad |x+\lambda h_1|
\end{aligned}
$$

**Шаг 4.** Разность $F(z+\lambda h)-F(z)$:
$$\Big(\lambda(yh_1+xh_2)+\lambda^2 h_1 h_2,\ \lambda(2xh_1+2yh_2)+\lambda^2(h_1^2+h_2^2),\ |x+\lambda h_1|-|x|\Big)$$

**Шаг 5.** Деление на $\lambda$:
$$\frac{F(z+\lambda h)-F(z)}{\lambda} = \left( yh_1+xh_2+\lambda h_1 h_2,\ 2xh_1+2yh_2+\lambda(h_1^2+h_2^2),\ \frac{|x+\lambda h_1|-|x|}{\lambda} \right)$$

**Шаг 6.** Предел при $\lambda \to 0$ — см. ниже.

### Случай $x \neq 0$

$$|x+\lambda h_1| = \operatorname{sgn}(x)\,(x+\lambda h_1) \quad \text{при малых } \lambda$$

$$\frac{|x+\lambda h_1|-|x|}{\lambda} = \operatorname{sgn}(x)\,h_1$$

**Шаг 7.** Вариация Лагранжа:
$$\delta F(z,h) = \big(yh_1+xh_2,\ 2xh_1+2yh_2,\ \operatorname{sgn}(x)\,h_1\big)$$

**Шаг 8.** Матричная запись:
$$\delta F(z,h) = \begin{pmatrix} y & x \\ 2x & 2y \\ \operatorname{sgn}(x) & 0 \end{pmatrix}\begin{pmatrix} h_1 \\ h_2\end{pmatrix} = F'_r(z)\,h$$

**Шаг 9.** Условие (2): матрица $3\times 2$ с конечными элементами задаёт ограниченный линейный оператор $\mathbb{R}^2 \to \mathbb{R}^3$.

**Шаг 10.** Условие (3): все элементы матрицы непрерывны по $z$ при $x\neq 0$; $\operatorname{sgn}(x)$ постоянен на $x>0$ и $x<0$.

**Ответ:**
$$\boxed{F'(z) = F'_r(z) = \begin{pmatrix} y & x \\ 2x & 2y \\ \operatorname{sgn}(x) & 0 \end{pmatrix}}$$

---

### Случай $x = 0$, $z_0=(0,y_0)$, $h=(1,0)$

$$\frac{|0+\lambda|-0}{\lambda} = \frac{|\lambda|}{\lambda} = \operatorname{sgn}(\lambda)$$

$$\lim_{\lambda \to 0^+}\operatorname{sgn}(\lambda) = 1, \qquad \lim_{\lambda \to 0^-}\operatorname{sgn}(\lambda) = -1$$

$$\Rightarrow \text{предела нет} \Rightarrow \text{нет } \delta F \Rightarrow \text{нет } F'_r \Rightarrow \text{нет } F'$$

**Ответ:**
$$\boxed{F \text{ не дифференцируема по Фреше в точках } (0,y_0)}$$

## Пункт 2

$$ (Fx)(t) = \left(\ddot x^2(t) - \cos x(t)\right)^2, \qquad F: C^2[0,1] \to C[0,1] $$

$$ \|h\|_{C^2} = \max|h| + \max|\dot h| + \max|\ddot h|, \qquad \|f\|_C = \max|f(t)| $$

**Шаг 1.** Обозначение: 
$$ G(x) = \ddot x^2 - \cos x \Rightarrow F(x) = G^2(x) $$

**Шаг 2.** Приращение внутренней функции:
$$ g_\lambda = G(x + \lambda h) = (\ddot x + \lambda \ddot h)^2 - \cos(x + \lambda h) $$

**Шаг 3.** Раскрытие скобок:
$$ (\ddot x + \lambda \ddot h)^2 = \ddot x^2 + 2\lambda \ddot x \ddot h + \lambda^2 \ddot h^2 $$
$$ \cos(x + \lambda h) = \cos x \cos(\lambda h) - \sin x \sin(\lambda h) $$

**Шаг 4.** Разность \$g_\lambda - G\$:
$$
\begin{aligned}
g_\lambda - G &= 2\lambda \ddot x \ddot h + \lambda^2 \ddot h^2 + \cos x - \cos(x + \lambda h) \\
&= 2\lambda \ddot x \ddot h + \lambda^2 \ddot h^2 + \cos x (1 - \cos(\lambda h)) + \sin x \sin(\lambda h)
\end{aligned}
$$

**Шаг 5.** Вынесение \$\lambda\$:
$$ g_\lambda - G = \lambda \cdot \Psi_\lambda $$
$$ \Psi_\lambda = 2\ddot x\ddot h + \lambda \ddot h^2 + \cos x \cdot \frac{1 - \cos(\lambda h)}{\lambda} + \sin x \cdot \frac{\sin(\lambda h)}{\lambda} $$

**Шаг 6.** Предел \$\Psi_\lambda\$ при \$\lambda \to 0\$:
$$
\begin{aligned}
& \frac{1 - \cos(\lambda h)}{\lambda} = \frac{1 - \cos(\lambda h)}{\lambda h} \cdot h \to 0 \cdot h = 0 \\
& \frac{\sin(\lambda h)}{\lambda} = \frac{\sin(\lambda h)}{\lambda h} \cdot h \to 1 \cdot h = h \\
& \Rightarrow \Psi = \lim_{\lambda \to 0} \Psi_\lambda = 2\ddot x\ddot h + h \sin x
\end{aligned}
$$

**Шаг 7.** Оценка скорости сходимости:
$$
\begin{aligned}
& |1 - \cos u| \le \frac{u^2}{2}, \quad |\sin u - u| \le \frac{|u|^3}{6} \\
& \Rightarrow \|\Psi_\lambda - \Psi\|_C \le |\lambda| \left( \|\ddot h\|_C^2 + \frac{1}{2}\|h\|_C^2 + \frac{1}{6}\|h\|_C^3 \right) \to 0
\end{aligned}
$$

**Шаг 8.** Вариация \$F = G^2\$ (вычисление предела разностного отношения):
$$
\begin{aligned}
\frac{F(x + \lambda h) - F(x)}{\lambda} &= \frac{g_\lambda^2 - G^2}{\lambda} = \frac{(g_\lambda - G)(g_\lambda + G)}{\lambda} \\
&= \Psi_\lambda \cdot (g_\lambda + G) = \Psi_\lambda \cdot (2G + \lambda \Psi_\lambda)
\end{aligned}
$$

**Шаг 9.** Переход к пределу:
$$ \lim_{\lambda \to 0} \Psi_\lambda \cdot (2G + \lambda \Psi_\lambda) = \Psi \cdot 2G = 2G \cdot \Psi $$

**Шаг 10.** Подстановка:
$$ \delta F(x, h) = 2(\ddot x^2 - \cos x)(2\ddot x\ddot h + h \sin x) $$

**Шаг 11.** Проверка условия (2) — линейность и ограниченность:
$$
\begin{aligned}
& \delta F \text{ линейна по } h \\
& \|\delta F(x, h)\|_C \le 2\|G\|_C \left( 2\|\ddot x\|_C \|\ddot h\|_C + \|h\|_C \right) \\
& \|\ddot h\|_C \le \|h\|_{C^2}, \quad \|h\|_C \le \|h\|_{C^2} \\
& \Rightarrow \|\delta F(x, h)\|_C \le 2\|G\|_C (2\|\ddot x\|_C + 1) \|h\|_{C^2} \\
& \Rightarrow F'_r(x) \in \mathcal{L}(C^2, C), \quad F'_r(x)h = \delta F(x, h)
\end{aligned}
$$

**Шаг 12.** Проверка условия (3) — непрерывность $F'_r$ по $x$. 
Пусть $\tilde{x} = x + k$, $\tilde{G} = G(\tilde{x})$:
$$
\begin{aligned}
F'_r(\tilde{x})h - F'_r(x)h &= 2(\tilde{G} - G)(2\ddot{\tilde{x}}\ddot{h} + h \sin \tilde{x}) \\
&\quad + 2G(2\ddot{k}\ddot{h} + h(\sin \tilde{x} - \sin x))
\end{aligned}
$$

**Шаг 13.** Оценки разностей:
$$
\begin{aligned}
\tilde{G} - G &= 2\ddot{x}\ddot{k} + \ddot{k}^2 + \cos x - \cos(x + k) \\
\|\tilde{G} - G\|_C &\le (2\|\ddot{x}\|_C + \|k\|_{C^2} + 1)\|k\|_{C^2} \\
|\sin \tilde{x} - \sin x| &\le |k| \Rightarrow \|\sin \tilde{x} - \sin x\|_C \le \|k\|_{C^2} \\
\|\ddot{\tilde{x}}\|_C &\le \|\ddot{x}\|_C + \|k\|_{C^2}
\end{aligned}
$$

**Шаг 14.** Итоговая оценка:
$$
\begin{aligned}
\|F'_r(x + k) - F'_r(x)\| &\le 2(2\|\ddot{x}\|_C + \|k\|_{C^2} + 1)\|k\|_{C^2}(2\|\ddot{x}\|_C + 2\|k\|_{C^2} + 1) \\
&\quad + 6\|G\|_C \|k\|_{C^2} \xrightarrow{k \to 0} 0
\end{aligned}
$$
$$ \Rightarrow F'_r \text{ непрерывна в каждой точке } x \in C^2[0,1] $$
$$ \Rightarrow F \text{ дифференцируема по Фреше} $$

**Ответ:**
$$ \boxed{F'(x)h = F'_r(x)h = 2(\ddot x^2 - \cos x)(2\ddot x\ddot h + h \sin x)} $$

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import HTML
import warnings
from matplotlib import MatplotlibDeprecationWarning

warnings.simplefilter('ignore', MatplotlibDeprecationWarning)

plt.rcParams.update({
    'figure.dpi': 90,
    'font.size': 11,
    'axes.grid': True,
    'grid.alpha': 0.3,
})

OUT = 'anim'
os.makedirs(OUT, exist_ok=True)

def show(anim):
    return HTML(anim.to_jshtml())
```

```python
fig, ax = plt.subplots(figsize=(7.5, 5))
xg = np.linspace(-0.5, 2.5, 400)
ax.plot(xg, np.abs(xg), 'b-', lw=3, label=r'$f(x)=|x|$')
sec, = ax.plot([], [], 'r--', lw=2, label='секущая')
ax.plot([1], [1], 'ko', ms=10, zorder=5)
txt = ax.text(0.03, 0.95, '', transform=ax.transAxes, fontsize=12,
              va='top', bbox=dict(fc='white', ec='0.7'))
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_xlim(-0.5, 2.5); ax.set_ylim(-0.5, 3.0)
ax.set_xlabel('x'); ax.set_ylabel('y'); ax.legend(loc='lower right')
lams = np.linspace(-3.0, 3.0, 40)

def update1(i):
    lam = lams[i] if abs(lams[i]) > 1e-12 else 1e-12
    slope = (abs(1 + lam) - 1) / lam
    sec.set_data(xg, 1 + slope * (xg - 1))
    txt.set_text(f'λ = {lam:+.3f}\nнаклон = {slope:.4f}')
    ax.set_title(f'Пункт 1, $x_0=1$: секущие, λ = {lam:+.3f}')
    return sec, txt

anim1 = FuncAnimation(fig, update1, frames=len(lams), interval=120, blit=False)
plt.close(fig)
show(anim1)
```

```python
fig, ax = plt.subplots(figsize=(7.5, 5))
xg = np.linspace(-2, 2, 400)
ax.plot(xg, np.abs(xg), 'b-', lw=3, label=r'$f(x)=|x|$')
sec, = ax.plot([], [], 'r--', lw=2, label='секущая')
ax.plot([0], [0], 'ko', ms=10, zorder=5)
txt = ax.text(0.03, 0.95, '', transform=ax.transAxes, fontsize=12,
              va='top', bbox=dict(fc='white', ec='0.7'))
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_xlim(-2, 2); ax.set_ylim(-2.0, 2.5)
ax.set_xlabel('x'); ax.set_ylabel('y'); ax.legend(loc='lower right')
lams = np.linspace(-1.5, 1.5, 40)

def update2(i):
    lam = lams[i] if abs(lams[i]) > 1e-12 else 1e-12
    slope = abs(lam) / lam
    sec.set_data(xg, slope * xg)
    txt.set_text(f'λ = {lam:+.3f}\nнаклон = {slope:+.3f}')
    ax.set_title(f'Пункт 1, $x_0=0$: излом, λ = {lam:+.3f}')
    return sec, txt

anim2 = FuncAnimation(fig, update2, frames=len(lams), interval=120, blit=False)
plt.close(fig)
show(anim2)
```

```python
fig, ax = plt.subplots(figsize=(7.5, 5))
tneg = np.linspace(-1, 0, 300, endpoint=False)
tpos = np.linspace(0, 1, 300)
ax.plot(tneg, -np.ones_like(tneg), 'r-', lw=3,
        label=r'$x_0=0:\ \varphi(\lambda)=\mathrm{sgn}(\lambda)$')
ax.plot(tpos, np.ones_like(tpos), 'r-', lw=3)
ax.plot([0, 0], [-1, 1], 'r:', lw=1)
ax.plot([-1, 1], [1, 1], 'g-', lw=3, label=r'$x_0=1:\ \varphi(\lambda)=1$')
p0, = ax.plot([], [], 'ro', ms=12)
p1, = ax.plot([], [], 'go', ms=12)
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_xlim(-1, 1); ax.set_ylim(-1.6, 1.6)
ax.set_xlabel('λ'); ax.set_ylabel('φ'); ax.legend(loc='center left')
lams = np.linspace(-1.0, 1.0, 40)

def update3(i):
    lam = lams[i] if abs(lams[i]) > 1e-12 else 1e-12
    p0.set_data([lam], [np.sign(lam)])
    p1.set_data([lam], [1.0])
    ax.set_title(f'φ(λ) при λ = {lam:+.3f}')
    return p0, p1

anim3 = FuncAnimation(fig, update3, frames=len(lams), interval=120, blit=False)
plt.close(fig)
show(anim3)
```

```python
ts = np.linspace(0, 1, 600)
xf  = lambda t: t**2
hf  = lambda t: np.sin(np.pi * t)
xdd = lambda t: 2.0 * np.ones_like(t)
hdd = lambda t: -(np.pi**2) * np.sin(np.pi * t)
Psi  = lambda t: 2 * xdd(t) * hdd(t) + hf(t) * np.sin(xf(t))
def PsiL(t, lam):
    return (2 * xdd(t) * hdd(t)
            + lam * hdd(t)**2
            + np.cos(xf(t)) * (1 - np.cos(lam * hf(t))) / lam
            + np.sin(xf(t)) * np.sin(lam * hf(t)) / lam)

lams = np.linspace(0.5, 0.02, 40)
all_vals = np.concatenate([PsiL(ts, l) for l in lams] + [Psi(ts)])
ymin, ymax = all_vals.min(), all_vals.max()
pad = 0.1 * (ymax - ymin)

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(ts, Psi(ts), 'k-', lw=4, label=r'$\Psi(t)$ — предел')
ln, = ax.plot([], [], 'r--', lw=2, label=r'$\Psi_\lambda(t)$')
ax.set_xlim(0, 1); ax.set_ylim(ymin - pad, ymax + pad)
ax.set_xlabel('t'); ax.set_ylabel('Ψ'); ax.legend(loc='lower right')

def update5(i):
    lam = lams[i]
    ln.set_data(ts, PsiL(ts, lam))
    ax.set_title(f'Пункт 2, шаг 6: λ = {lam:.4f}')
    return ln,

anim5 = FuncAnimation(fig, update5, frames=len(lams), interval=120, blit=False)
plt.close(fig)
show(anim5)
```

```python
tgrid = np.linspace(0, 1, 201)
xf  = lambda t: t**2
hf  = lambda t: np.sin(np.pi * t)
xdd = lambda t: 2.0 * np.ones_like(t)
hdd = lambda t: -(np.pi**2) * np.sin(np.pi * t)
Psi  = lambda t: 2 * xdd(t) * hdd(t) + hf(t) * np.sin(xf(t))
def PsiL(t, lam):
    return (2 * xdd(t) * hdd(t)
            + lam * hdd(t)**2
            + np.cos(xf(t)) * (1 - np.cos(lam * hf(t))) / lam
            + np.sin(xf(t)) * np.sin(lam * hf(t)) / lam)

lams = np.linspace(0.5, 0.01, 40)
errs = np.array([np.max(np.abs(PsiL(tgrid, l) - Psi(tgrid))) for l in lams])
Cbound = (np.max(np.abs(hdd(tgrid)))**2
          + 0.5 * np.max(np.abs(hf(tgrid)))**2
          + (1/6) * np.max(np.abs(hf(tgrid)))**3)

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(lams, Cbound * lams, 'b--', lw=2,
        label=fr'оценка $C|\lambda|$, $C={Cbound:.1f}$')
pt, = ax.plot([], [], 'ro', ms=12)
vl, = ax.plot([], [], 'r:', lw=1)
hl, = ax.plot([], [], 'r-', lw=2, alpha=0.4)
ax.set_xlim(0, 0.5)
ax.set_ylim(0, max(Cbound * 0.5, errs.max()) * 1.15)
ax.set_xlabel('λ'); ax.set_ylabel(r'$\|\Psi_\lambda-\Psi\|_C$')
ax.legend(loc='upper left')

def update6(i):
    lam = lams[i]; e = errs[i]
    pt.set_data([lam], [e])
    vl.set_data([lam, lam], [0, e])
    hl.set_data([0, lam], [e, e])
    ax.set_title(f'Шаг 7: λ = {lam:.4f}, ошибка = {e:.3e}')
    return pt, vl, hl

anim6 = FuncAnimation(fig, update6, frames=len(lams), interval=120, blit=False)
plt.close(fig)
show(anim6)
```

```python
ts = np.linspace(0, 1, 400)
xf  = lambda t: t**2
hf  = lambda t: np.sin(np.pi * t)
xdd = lambda t: 2.0 * np.ones_like(t)
hdd = lambda t: -(np.pi**2) * np.sin(np.pi * t)
G    = lambda t: xdd(t)**2 - np.cos(xf(t))
Psi  = lambda t: 2 * xdd(t) * hdd(t) + hf(t) * np.sin(xf(t))
gL   = lambda t, lam: (xdd(t) + lam * hdd(t))**2 - np.cos(xf(t) + lam * hf(t))
Fq   = lambda t, lam: (gL(t, lam)**2 - G(t)**2) / lam
dF   = lambda t: 2 * G(t) * Psi(t)

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(ts, dF(ts), 'k-', lw=4, label=r'$\delta F(x,h) = 2G\Psi$')
ln, = ax.plot([], [], 'r--', lw=2, label='разностное отношение')
ax.set_xlim(0, 1)
ymax = np.max(np.abs(dF(ts))) * 1.35
ax.set_ylim(-ymax, ymax)
ax.set_xlabel('t'); ax.set_ylabel('значение'); ax.legend(loc='lower right')
lams = np.linspace(0.5, 0.02, 40)

def update7(i):
    lam = lams[i]
    ln.set_data(ts, Fq(ts, lam))
    ax.set_title(f'Пункт 2, шаг 10: λ = {lam:.4f}')
    return ln,

anim7 = FuncAnimation(fig, update7, frames=len(lams), interval=120, blit=False)
plt.close(fig)
show(anim7)
```

```python
tgrid = np.linspace(0, 1, 201)
xf  = lambda t: t**2
xdd = lambda t: 2.0 * np.ones_like(t)
G    = lambda t: xdd(t)**2 - np.cos(xf(t))
C_op = (2 * (2 * np.max(np.abs(xdd(tgrid))) + 1)**2
        + 6 * np.max(np.abs(G(tgrid))))

fig, ax = plt.subplots(figsize=(8, 5))
ep = np.linspace(0.01, 0.5, 100)

# ⬇️ БЫЛО: \le  →  СТАЛО: \leq  ;  \prime → '
ax.plot(ep, C_op * ep, 'r-', lw=3,
        label=r"$\|F_r'(x+k)-F_r'(x)\| \leq C\varepsilon$")
ax.plot([0.01, 0.5], [0, 0], 'k:', lw=2, label='предел = 0')

pt, = ax.plot([], [], 'bo', ms=12)
ax.set_xlim(0, 0.5)
ax.set_ylim(0, C_op * 0.5 * 1.15)
ax.set_xlabel(r'$\varepsilon = \|k\|$')
ax.set_ylabel('операторная норма')
ax.legend(loc='upper left')

epsilons = np.linspace(0.5, 0.01, 40)

def update8(i):
    e = epsilons[i]
    pt.set_data([e], [C_op * e])
    ax.set_title(f'Пункт 2, шаг 14: ε = ||k|| = {e:.4f}, C = {C_op:.1f}')
    return pt,

anim8 = FuncAnimation(fig, update8, frames=len(epsilons), interval=120, blit=False)
plt.close(fig)
show(anim8)
```
