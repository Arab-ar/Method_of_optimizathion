# Задача 17. Задача быстродействия

**Вариант 17**

$$T \to \inf, \quad -2 \le \ddot{x} \le 1, \quad x(0) = 1, \quad \dot{x}(0) = -2, \quad \dot{x}(T) = 1, \quad x(T) = 0.$$

## 1. Приведение к задаче оптимального управления

Вводим фазовые переменные $x_1 = x$, $x_2 = \dot{x}$ и управление $u = \ddot{x}$:

$$T \to \inf; \quad \dot{x}_1 = x_2, \quad \dot{x}_2 = u, \quad u \in [-2, 1],$$

$$x_1(0) = 1, \quad x_2(0) = -2, \quad x_1(T) = 0, \quad x_2(T) = 1.$$

## 2. Функция Лагранжа

$$\mathscr{L} = \int_0^T \big(p_1(\dot{x}_1 - x_2) + p_2(\dot{x}_2 - u)\big)\,dt + \lambda_0 T + \lambda_1(x_1(0) - 1) + \lambda_2(x_2(0) + 2) + \lambda_3 x_1(T) + \lambda_4(x_2(T) - 1).$$

## 3. Необходимые условия оптимальности

### а) Уравнения Эйлера для сопряжённых переменных

$$-\frac{d}{dt}\hat{L}_{\dot{x}_i} + \hat{L}_{x_i} = 0, \quad i = 1,2 \implies \dot{p}_1 = 0, \quad \dot{p}_2 = -p_1.$$

Отсюда:

$$p_1(t) = C_1, \quad p_2(t) = C_2 - C_1 t.$$

### б) Условия трансверсальности

$$p_1(0) = \lambda_1, \quad p_2(0) = \lambda_2, \quad p_1(T) = -\lambda_3, \quad p_2(T) = -\lambda_4.$$

### в) Оптимальность по управлению $u$

$$\min_{u \in [-2, 1]} (-p_2(t)u) = -p_2(t)\hat{u}(t) \implies \hat{u}(t) = \begin{cases} 1, & p_2(t) > 0, \\ -2, & p_2(t) < 0. \end{cases}$$

### г) Стационарность по $T$

$$\mathscr{L}_T = 0 \implies \lambda_0 - p_1(T)x_2(T) - p_2(T)\hat{u}(T) = 0.$$

Поскольку $x_2(T) = 1$:

$$\lambda_0 = p_1(T) + p_2(T)\hat{u}(T).$$

## 4. Обоснование нетривиальности: $\lambda_0 \ne 0$

Предположим $\lambda_0 = 0$. Тогда $p_1(T) = -p_2(T)\hat{u}(T)$.

**Случай 1:** $p_2(T) > 0 \implies \hat{u}(T) = 1 \implies p_1(T) = -p_2(T) \implies p_2(t) = C(t - T + 1)$.

Корень $p_2$: $\tau = T - 1$. Если $\tau \notin (0, T)$, то $\hat{u} \equiv 1$.

Изменение скорости: $\Delta x_2 = x_2(T) - x_2(0) = 1 - (-2) = 3$.

При $\hat{u} \equiv 1$: $\Delta x_2 = T \implies T = 3 \implies \tau = 2$.

Проверка $x_1(T)$:

$$x_1(T) = 1 - 2 \cdot 3 + \int_0^2 (3-t)(-2)\,dt + \int_2^3 (3-t)(1)\,dt = 1 - 6 - 8 + 0.5 = -12.5 \ne 0.$$

**Противоречие.**

**Случай 2:** $p_2(T) < 0 \implies \hat{u}(T) = -2 \implies p_2(t) < 0$ на $[0, T] \implies \hat{u} \equiv -2$.

Тогда $\dot{x}_2 < 0 \implies x_2(T) < x_2(0) = -2$, но нужно $x_2(T) = 1$. **Противоречие.**

**Вывод:** $\lambda_0 \ne 0$, нормируем $\lambda_0 = 1$.

## 5. Выбор структуры управления

$p_2(t) = C_2 - C_1 t$ — линейна $\implies$ **не более одного корня** $\implies$ не более одного переключения.

Поскольку $x_2(0) = -2 < x_2(T) = 1$, нужно $\Delta x_2 > 0$.

- Структура $u^+: \{-2 \to 1\}$: $\dot{x}_2(0) = -2 \implies x_2$ убывает $\implies$ недостижимо.
- Структура $u^-: \{1 \to -2\}$: $\dot{x}_2(0) = 1 \implies x_2$ возрастает $\implies$ достижимо.

$$\boxed{u^*(t) = \begin{cases} 1, & 0 \le t < \tau, \\ -2, & \tau \le t \le T. \end{cases}}$$

## 6. Фазовые траектории и линия переключения Г

**При $u = 1$:** $\dot{x}_2 = 1 \implies x_2 = t + C_1 \implies x_1 = \frac{x_2^2}{2} + C$.

**При $u = -2$:** $\dot{x}_2 = -2 \implies x_2 = -2t + C_1 \implies x_1 = -\frac{x_2^2}{4} + C$.

Условие попадания в $(0, 1)$ задаёт линию переключения:

$$Г: x_1 = \begin{cases} -\dfrac{x_2^2}{4} + \dfrac{1}{4}, & x_2 \ge 1, \\[6pt] \dfrac{x_2^2}{2} - \dfrac{1}{2}, & x_2 \le 1. \end{cases}$$

## 7. Траектория из начальной точки $(1, -2)$

При $u = 1$: $(x_1, x_2) = (1, -2) \implies 1 = \frac{(-2)^2}{2} + C \implies C = -1$.

$$x_1 = \frac{x_2^2}{2} - 1.$$

Точка переключения — пересечение с веткой $Г$ при $x_2 \ge 1$:

$$\frac{x_2^2}{2} - 1 = -\frac{x_2^2}{4} + \frac{1}{4} \implies \frac{3}{4}x_2^2 = \frac{5}{4} \implies x_2(\tau) = \sqrt{\frac{5}{3}}.$$

$$x_1(\tau) = \frac{1}{2} \cdot \frac{5}{3} - 1 = -\frac{1}{6} \implies M_\tau\left(-\frac{1}{6},\, \sqrt{\frac{5}{3}}\right).$$

## 8. Вычисление времени переключения и полного времени

**На $[0, \tau]$:** $x_2(t) = -2 + t \implies \sqrt{5/3} = -2 + \tau \implies \tau = 2 + \sqrt{5/3}$.

**На $[\tau, T]$:** $x_2(t) = \sqrt{5/3} - 2(t - \tau) \implies 1 = \sqrt{5/3} - 2(T - \tau)$.

$$T - \tau = \frac{\sqrt{5/3} - 1}{2}.$$

$$T_{\min} = \tau + (T - \tau) = 2 + \sqrt{\frac{5}{3}} + \frac{\sqrt{5/3} - 1}{2} = \frac{3}{2}\left(1 + \sqrt{\frac{5}{3}}\right) \approx 3.436.$$

## Ответ

$$u^*(t) = \begin{cases} 1, & 0 \le t < 2 + \sqrt{5/3}, \\ -2, & 2 + \sqrt{5/3} \le t \le T_{\min}. \end{cases}$$

$$T_{\min} = \frac{3}{2}\left(1 + \sqrt{\frac{5}{3}}\right) \approx 3.436.$$
