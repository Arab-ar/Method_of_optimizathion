# Задача 3. Касательное пространство к множеству $M$ в точке $(0,0)$

$$M = \{(x_1, x_2) \in \mathbb{R}^2 \mid x_1 \le 0,\ x_2 \le 0,\ x_1^3 + x_2^3 \le 1\}, \qquad p = (0, 0).$$

## Часть I. Теоретические основания

### 1.1. Что такое касательный вектор?

**Определение.** Касательный вектор в точке $p$ — это **вектор скорости** $\alpha'(0)$ гладкой кривой $\alpha$, целиком лежащей в $M$ и проходящей через $p$:

$$\alpha: (-\varepsilon, \varepsilon) \to M, \quad \alpha(0) = p.$$

Две кривые $\alpha$ и $\beta$ задают один касательный вектор, если $\alpha'(0) = \beta'(0)$.

> **Замечание.** Для множеств с границей кривая определяется на $[0, \varepsilon)$, так как из граничной точки нельзя двигаться «наружу».

### 1.2. Касательный конус и пространство

**Касательный конус** — множество всех возможных векторов скорости:

$$K_M(p) = \{ v \in \mathbb{R}^n \mid \exists\, \alpha: [0,\varepsilon) \to M,\ \alpha(0) = p,\ \alpha'(0) = v \}.$$

**Касательное пространство** — линейная оболочка конуса:

$$T_p M = \text{span}\, K_M(p) = \left\{ \sum_{i=1}^k \lambda_i v_i \;\middle|\; k \in \mathbb{N},\ \lambda_i \in \mathbb{R},\ v_i \in K_M(p) \right\}.$$

> **Замечание.** Для открытого $U \subset \mathbb{R}^n$: $T_p U \cong \mathbb{R}^n$. Для множества с границей $K_M(p)$ может быть строго меньше $\mathbb{R}^n$, но $\text{span}\, K_M(p)$ всё равно может дать $\mathbb{R}^n$.

## Часть II. Решение

### Шаг 1. Анализ ограничений

### Шаг 2. Необходимое условие

Пусть $\alpha: [0, \varepsilon) \to M$, $\alpha(0) = (0,0)$, $\alpha'(0) = v = (v_1, v_2)$.

Так как $\alpha(t) \in M$, то $\alpha_1(t) \le 0,\ \alpha_2(t) \le 0$ для всех $t \in [0, \varepsilon)$. Отсюда:

$$v_1 = \alpha_1'(0) = \lim_{t \to 0^+} \frac{\alpha_1(t)}{t} \le 0, \quad v_2 = \alpha_2'(0) = \lim_{t \to 0^+} \frac{\alpha_2(t)}{t} \le 0.$$

### Шаг 3. Достаточное условие

Для любого $v = (v_1, v_2)$ с $v_1 \le 0,\ v_2 \le 0$ рассмотрим кривую $\alpha(t) = (tv_1, tv_2)$, $t \in [0, \varepsilon)$:

$$\alpha_1(t) = tv_1 \le 0, \quad \alpha_2(t) = tv_2 \le 0, \quad \alpha_1^3 + \alpha_2^3 = t^3(v_1^3 + v_2^3) \le 0 < 1.$$

Значит $\alpha(t) \in M$, $\alpha(0) = p$, $\alpha'(0) = v \Rightarrow v \in K_M(p)$.

### Шаг 4. Касательный конус

$$K_M(0,0) = \{(v_1, v_2) \in \mathbb{R}^2 \mid v_1 \le 0,\ v_2 \le 0\} \quad \text{(третий квадрант)}.$$

### Шаг 5. Касательное пространство

В конусе содержатся $u_1 = (-1, 0)$ и $u_2 = (0, -1)$. Система $\{u_1, u_2\}$ — базис $\mathbb{R}^2$, поэтому:

$$T_{(0,0)}M = \text{span}\, K_M(0,0) \supseteq \text{span}\{u_1, u_2\} = \mathbb{R}^2.$$

## Ответ

$$\boxed{T_{(0,0)}M = \mathbb{R}^2}$$

Точка $(0,0)$ — «угол» множества $M$ (пересечение прямых $x_1=0$ и $x_2=0$). Касательный конус — третий квадрант. Хотя $K_M(0,0) \neq \mathbb{R}^2$, он содержит два линейно независимых вектора $(-1,0)$ и $(0,-1)$, поэтому его линейная оболочка заполняет всю плоскость.
