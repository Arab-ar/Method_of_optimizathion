# Касательное пространство $T_{(0,0)}M$

$$
M = \{(x_1, x_2) \in \mathbb{R}^2 \mid x_1 \le 0,\ x_2 \le 0,\ x_1^3 + x_2^3 \le 1\}, \qquad x_0 = (0, 0)
$$

## 1. Ограничения и их активность в $x_0$

$$
g_1(x_1, x_2) = x_1 \le 0 \;\Rightarrow\; g_1(0, 0) = 0 \quad (\text{активно})
$$

$$
g_2(x_1, x_2) = x_2 \le 0 \;\Rightarrow\; g_2(0, 0) = 0 \quad (\text{активно})
$$

$$
g_3(x_1, x_2) = x_1^3 + x_2^3 - 1 \le 0 \;\Rightarrow\; g_3(0, 0) = -1 < 0 \quad (\text{неактивно})
$$

Для любой точки $x \in M$:
$$
x_1 \le 0,\ x_2 \le 0 \;\Rightarrow\; x_1^3 + x_2^3 \le 0 < 1
$$

$$
\Rightarrow\; I(x_0) = \{1, 2\}
$$

## 2. Градиенты и регулярность

$$
\nabla g_1(x_0) = \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \qquad \nabla g_2(x_0) = \begin{pmatrix} 0 \\ 1 \end{pmatrix}
$$

$$
\det \begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix} = 1 \neq 0
$$

$\Rightarrow$ условие регулярности выполнено.

## 3. Допустимые направления

Условие: $\nabla g_i(x_0)^T v \le 0,\ i \in I(x_0)$

$$
\nabla g_1(0, 0)^T v = \begin{pmatrix} 1 & 0 \end{pmatrix} \begin{pmatrix} v_1 \\ v_2 \end{pmatrix} = v_1 \le 0
$$

$$
\nabla g_2(0, 0)^T v = \begin{pmatrix} 0 & 1 \end{pmatrix} \begin{pmatrix} v_1 \\ v_2 \end{pmatrix} = v_2 \le 0
$$

$$
D = \left\{ v = (v_1, v_2) \in \mathbb{R}^2 \;\middle|\; v_1 \le 0,\ v_2 \le 0 \right\}
$$

## 4. Касательное пространство

Касательное пространство — линейная оболочка допустимых направлений:

$$
T_{(0,0)}M = \operatorname{span}(D) = \left\{ \alpha u + \beta w \mid \alpha, \beta \in \mathbb{R},\ u, w \in D \right\}
$$

Так как $(-1, 0) \in D$ и $(0, -1) \in D$, то для любой точки $(x, y) \in \mathbb{R}^2$:

$$
(x, y) = -x(-1, 0) - y(0, -1) \;\Rightarrow\; \mathbb{R}^2 \subseteq \operatorname{span}(D)
$$

$$
\Rightarrow\; T_{(0,0)}M = \mathbb{R}^2
$$

## Ответ

$$
T_{(0,0)}M = \mathbb{R}^2
$$
