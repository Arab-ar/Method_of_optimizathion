# Задача: Касательный конус $T_{x_0}M$ (Вариант 17)

**Условие задачи:**
Дано множество и точка:
$$
M = \{(x_1, x_2) \in \mathbb{R}^2 \mid x_1 \le 0,\ x_2 \le 0,\ x_1^3 + x_2^3 \le 1\}, \qquad x_0 = (0, 0).
$$
Требуется найти касательный конус $T_{x_0}M$.

## 1. Ограничения и их активность в $x_0$

Запишем множество $M$ через систему неравенств $g_i(x) \le 0$:
$$
\begin{aligned}
g_1(x_1, x_2) &= x_1 \le 0 &&\implies g_1(0, 0) = 0 \quad (\text{активно}), \\
g_2(x_1, x_2) &= x_2 \le 0 &&\implies g_2(0, 0) = 0 \quad (\text{активно}), \\
g_3(x_1, x_2) &= x_1^3 + x_2^3 - 1 \le 0 &&\implies g_3(0, 0) = 0^3 + 0^3 - 1 = -1 < 0 \quad (\text{неактивно}).
\end{aligned}
$$

Так как для любых $x \in M$ выполняется $x_1 \le 0,\ x_2 \le 0$, то $x_1^3 + x_2^3 \le 0 < 1$. Следовательно, индексное множество активных ограничений:
$$
I(x_0) = \{1, 2\}.
$$

## 2. Градиенты и регулярность

Вычислим градиенты активных ограничений в точке $x_0$:
$$
\nabla g_1(x_0) = \left. \begin{pmatrix} \dfrac{\partial g_1}{\partial x_1} \\[8pt] \dfrac{\partial g_1}{\partial x_2} \end{pmatrix} \right|_{(0,0)} = \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \qquad
\nabla g_2(x_0) = \left. \begin{pmatrix} \dfrac{\partial g_2}{\partial x_1} \\[8pt] \dfrac{\partial g_2}{\partial x_2} \end{pmatrix} \right|_{(0,0)} = \begin{pmatrix} 0 \\ 1 \end{pmatrix}.
$$

Проверим условие регулярности (линейную независимость градиентов):
$$
\det \begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix} = 1 \neq 0 \implies \text{условие регулярности выполнено.}
$$

## 3. Условие касательного конуса

Касательный конус задается системой линейных неравенств: $\nabla g_i(x_0)^T v \le 0$ для всех $i \in I(x_0)$.
$$
\begin{aligned}
\nabla g_1(0, 0)^T v &= \begin{pmatrix} 1 & 0 \end{pmatrix} \begin{pmatrix} v_1 \\ v_2 \end{pmatrix} = v_1 \le 0, \\[4pt]
\nabla g_2(0, 0)^T v &= \begin{pmatrix} 0 & 1 \end{pmatrix} \begin{pmatrix} v_1 \\ v_2 \end{pmatrix} = v_2 \le 0.
\end{aligned}
$$

### Ответ
$$
T_{(0,0)}M = \left\{ v = (v_1, v_2) \in \mathbb{R}^2 \;\middle|\; v_1 \le 0,\ v_2 \le 0 \right\}.
$$
