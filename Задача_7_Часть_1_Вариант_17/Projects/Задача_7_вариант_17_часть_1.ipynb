# Задача: Экстремумы многочлена нечетной степени (Задача 7)

**Условие задачи:**
Пусть $P$ — многочлен нечетной степени, имеющий две точки $x_1, x_2$ такие, что $P'(x_1) = P'(x_2) = 0$. Будут ли точки $x_1, x_2$ точками экстремума?

**Ответ:** Нет, не обязательно (при $\deg P \ge 5$).

## 1. Случай $\deg P = 3$ (экстремумы всегда есть)

$$
\deg P = 3 \implies \deg P' = 2
$$

$$
P'(x_1) = P'(x_2) = 0 \ (x_1 \neq x_2) \implies x_1, x_2 \text{ --- простые корни (кратности 1)}
$$

$$
\implies P'(x) \text{ меняет знак при переходе через } x_1 \text{ и } x_2 \implies \text{обе точки являются экстремумами}
$$

## 2. Контрпример для $\deg P = 5$

$$
P(x) = \frac{x^5}{5} - \frac{2x^3}{3} + x \quad (\deg P = 5 \text{ --- нечетная})
$$

$$
P'(x) = x^4 - 2x^2 + 1 = (x^2 - 1)^2 = (x-1)^2(x+1)^2
$$

$$
P'(x) = 0 \implies x_1 = -1, \quad x_2 = 1
$$

$$
P'(x) \ge 0 \ \forall x \in \mathbb{R} \implies \text{знак не меняется} \implies \text{экстремумов нет}
$$

## 3. Проверка контрпримера через производные высших порядков

$$
P''(x) = 4x^3 - 4x = 4x(x^2 - 1) \implies P''(\pm 1) = 0
$$

$$
P'''(x) = 12x^2 - 4 \implies P'''(\pm 1) = 8 \neq 0
$$

$$
\implies x_1, x_2 \text{ --- точки перегиба}
$$

## 4. Общий случай для произвольной нечетной степени $\deg P = 2k+1 \ge 5$

$$
\deg P = 2k+1 \ (k \ge 2) \implies \deg P' = 2k
$$

$$
\text{Пусть } P'(x) = a(x - x_1)^{m_1}(x - x_2)^{m_2}, \text{ где } m_1, m_2 \text{ --- четные, } m_1 + m_2 = 2k
$$

$$
\implies P'(x) \text{ не меняет знак при переходе через } x_1, x_2 \implies \text{экстремумов нет}
$$
