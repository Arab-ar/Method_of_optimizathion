# Задача 9 (Часть 2) — Вариант 17

Найти экстремум функционала:

$$J[x] = \int_0^1 \dot{x}^2\,dt \to \operatorname{extr},$$

при ограничении:

$$g(x) := \int_0^1 x^2\,dt - 1 \ge 0, \qquad x(0) = x(1) = 0.$$

## 1. Необходимые условия (KKT)

Лагранжиан для $g(x) \ge 0$:

$$\mathcal{L}[x,\mu] = J[x] - \mu\, g(x) = \int_0^1 \bigl(\dot{x}^2 - \mu\, x^2\bigr)\,dt + \mu, \quad \mu \ge 0,$$

$$\mu\left(\int_0^1 x^2\,dt - 1\right) = 0.$$

Первая вариация:

$$\delta \mathcal{L} = 2\int_0^1 \bigl(\dot{x}\,\delta\dot{x} - \mu\, x\,\delta x\bigr)\,dt = -2\int_0^1 \bigl(\ddot{x} + \mu\, x\bigr)\delta x\,dt.$$

Учитывая $\delta x(0) = \delta x(1) = 0$, получаем уравнение Эйлера–Лагранжа:

$$\ddot{x} + \mu\, x = 0.$$

## 2. Обоснование активности ограничения

**Предположим, что ограничение неактивно:** $\int_0^1 x^2\,dt > 1$.

Тогда по условию дополняющей нежёсткости $\mu = 0$, и уравнение Эйлера даёт:

$$\ddot{x} = 0, \quad x(0) = x(1) = 0 \;\Rightarrow\; x \equiv 0.$$

Но тогда $\int_0^1 x^2\,dt = 0 < 1$, что противоречит допустимости $x$.

$\Rightarrow$ **ограничение активно:** $\int_0^1 x^2\,dt = 1$, $\mu > 0$.

## 3. Сведение к задаче Лагранжа

Так как ограничение активно, вводим замену $u = \dot{x}$ и переходим к задаче с дифференциальной связью:

$$J = \int_0^1 u^2\,dt \to \operatorname{extr}, \qquad \dot{x} - u = 0, \quad x(0) = x(1) = 0, \quad \int_0^1 x^2\,dt = 1.$$

**Функция Лагранжа:**

$$L = \lambda_0\, u^2 + p(t)(\dot{x} - u) + \lambda\, x^2.$$

**Интегрант:**

$$\mathcal{L}(t, x, \dot{x}, u, p, \lambda_0, \lambda) = \lambda_0\, u^2 + p(t)(\dot{x} - u) + \lambda\, x^2.$$

**Терминант:**

$$l(x(0), x(1)) = 0 \quad \text{(отсутствует, так как концы фиксированы: } x(0) = x(1) = 0\text{)}.$$

Множители: $\lambda_0$, $\lambda$ и $p(t)$.

## 4. Необходимые условия

1. **Уравнения Эйлера:**

$$\frac{d}{dt} L_{\dot{x}} = L_x \;\Rightarrow\; \dot{p} = 2\lambda\, x.$$

2. **Трансверсальность по $x$:**

$$L_{\dot{x}}(t_k) = (-1)^k\, l_{x(t_k)}, \quad k = 0, 1.$$

Концы фиксированы: $x(0) = x(1) = 0 \;\Rightarrow\; \delta x(0) = \delta x(1) = 0 \;\Rightarrow\; \bigl[L_{\dot{x}}\,\delta x\bigr]_0^1 = 0$.

3. **Оптимальность по $u$:**

$$L_u = 0 \;\Rightarrow\; 2\lambda_0\, u - p = 0 \;\Rightarrow\; p = 2\lambda_0\, u.$$

$L_{uu} = 2\lambda_0 > 0$ при $\lambda_0 > 0 \;\Rightarrow\; u$ доставляет минимум $L$ по $u$.

4. **Согласование знаков:** $\lambda_0 \ge 0$; $\lambda$ любого знака.

## 5. Исследование нормальности

**Случай $\lambda_0 = 0$:**

$$p \equiv 0 \;\Rightarrow\; \lambda\, x \equiv 0.$$

Так как по ограничению $x \not\equiv 0$, то $\lambda = 0$, и все множители нулевые — невозможно.

**Случай $\lambda_0 \ne 0$:**

Если $\lambda_0 > 0$, делим все уравнения на $\lambda_0$ и переобозначаем $\lambda/\lambda_0$ как новую $\lambda$. Полагаем $\lambda_0 = 1$:

$$\lambda_0 = 1 \;\Rightarrow\; p = 2\dot{x}, \quad \dot{p} = 2\ddot{x}.$$

Из $\dot{p} = 2\lambda\, x$ получаем:

$$\ddot{x} = \lambda\, x, \qquad x(0) = x(1) = 0.$$

Усиленное условие Лежандра: $L_{uu} = 2 > 0$.

Кандидаты на экстремум — все $x_n(t) = \pm\sqrt{2}\sin(\pi n t)$, $n \in \mathbb{N}$.

## 6. Краевая задача

$$\ddot{x} - \lambda\, x = 0, \quad \lambda = -\omega^2 \;\Rightarrow\; \ddot{x} + \omega^2\, x = 0,$$

$$x(t) = A\cos\omega t + B\sin\omega t,$$

$$x(0) = 0 \;\Rightarrow\; A = 0 \;\Rightarrow\; x(t) = B\sin\omega t,$$

$$x(1) = 0 \;\Rightarrow\; B\sin\omega = 0,\ B \ne 0 \;\Rightarrow\; \omega = \pi n,\ n = 1, 2, \dots$$

$$\lambda = -\pi^2 n^2, \qquad x_n(t) = B\sin(\pi n t).$$

## 7. Достаточность

### 7.1. Неравенство Виртингера

Для $x \in C^1[0,1]$, $x(0) = x(1) = 0$:

$$\int_0^1 x^2\,dt \le \frac{1}{\pi^2} \int_0^1 \dot{x}^2\,dt, \qquad \text{равенство} \Leftrightarrow x(t) = C\sin(\pi t).$$

Оценка $J[x]$ через $\int_0^1 x^2\,dt$:

$$J[x] \ge \pi^2 \int_0^1 x^2\,dt \ge \pi^2,$$

$$J[x] = \pi^2 \;\Leftrightarrow\; \int_0^1 x^2\,dt = 1 \ \wedge\ x = C\sin(\pi t) \;\Rightarrow\; \frac{C^2}{2} = 1 \;\Rightarrow\; C = \pm\sqrt{2}.$$

### 7.2. Разложение $J[\hat{x} + h]$

Пусть $x(t) = \hat{x}(t) + h(t)$ — допустимая функция ($h(0) = h(1) = 0$, $\int_0^1 (\hat{x}+h)^2\,dt = 1$). Тогда:

$$J[x] = J[\hat{x}] + 2\int_0^1 \dot{\hat{x}}\,\dot{h}\,dt + \int_0^1 \dot{h}^2\,dt.$$

Интегрируя по частям и используя $\ddot{\hat{x}} = -\pi^2 \hat{x}$:

$$J[x] = J[\hat{x}] - \pi^2 \int_0^1 h^2\,dt + \int_0^1 \dot{h}^2\,dt.$$

По неравенству Виртингера $\int_0^1 \dot{h}^2\,dt \ge \pi^2 \int_0^1 h^2\,dt$, поэтому:

$$J[x] \ge J[\hat{x}] = \pi^2.$$

Равенство достигается только при $h \equiv 0$.

$\Rightarrow$ $\hat{x}_\pm(t) = \pm\sqrt{2}\sin(\pi t)$ — **строгий глобальный минимум**.

Глобального максимума нет: $\alpha \to +\infty \Rightarrow J[\alpha x] = \alpha^2 J[x] \to +\infty$.

## 8. Оптимальное управление

$$\hat{u}_\pm(t) = \dot{\hat{x}}_\pm(t) = \pm\sqrt{2}\,\pi\cos(\pi t).$$

Две оптимальные пары:

$$\bigl(\hat{x}_+,\hat{u}_+\bigr) = \bigl(\sqrt{2}\sin(\pi t),\ \sqrt{2}\pi\cos(\pi t)\bigr),$$

$$\bigl(\hat{x}_-,\hat{u}_-\bigr) = \bigl(-\sqrt{2}\sin(\pi t),\ -\sqrt{2}\pi\cos(\pi t)\bigr).$$

$J[\hat{x}_+] = J[\hat{x}_-] = \pi^2 \;\Rightarrow\;$ обе пары оптимальны.

## 9. Локальные экстремумы и седловые точки

**При $n = 1$:** $\hat{x}_\pm = \pm\sqrt{2}\sin(\pi t)$ — глобальный минимум $\Rightarrow$ локальный.

**При $n \ge 2$:** исследуем знак второй вариации на допустимых направлениях.

Вариация $x_\varepsilon = (1+\delta)x_n + \varepsilon\sin(\pi t)$, $\delta = -\tfrac{\varepsilon^2}{4} + O(\varepsilon^4)$:

$$J[x_\varepsilon] = (1+\delta)^2 J[x_n] + \varepsilon^2 J[\sin(\pi t)] = \pi^2 n^2 + \varepsilon^2 \pi^2\!\left(1 - \tfrac{n^2}{2}\right) + O(\varepsilon^4).$$

$n \ge 2 \Rightarrow 1 - \tfrac{n^2}{2} < 0 \Rightarrow J[x_\varepsilon] < J[x_n]$ при $\varepsilon \ne 0$ малых $\Rightarrow$ $x_n$ не локальный минимум.

Вариация $x_\varepsilon = (1+\delta)x_n + \varepsilon\sin(M\pi t)$, $M > n$:

$$J[x_\varepsilon] = \pi^2 n^2 + \varepsilon^2 \pi^2\!\left(M^2 - \tfrac{n^2}{2}\right) + O(\varepsilon^4) > J[x_n] \;\Rightarrow\; x_n \text{ не локальный максимум}.$$

$\Rightarrow$ $x_n = \pm\sqrt{2}\sin(\pi n t)$, $n \ge 2$ — **седловые точки**.

## Ответ

$$\hat{x}_\pm(t) = \pm\sqrt{2}\sin(\pi t), \qquad \hat{u}_\pm(t) = \pm\sqrt{2}\,\pi\cos(\pi t), \qquad J_{\min} = \pi^2.$$

Обе пары $(\hat{x}_\pm, \hat{u}_\pm)$ дают $J_{\min} = \pi^2 \;\Rightarrow\;$ обе оптимальны. Глобального максимума нет; $x_n$, $n \ge 2$ — седловые точки.
